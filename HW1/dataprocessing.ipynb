```python
import pandas as pd
```

```python
data = pd.read_parquet(
    'dataset/green_tripdata_2021-01.parquet',
    engine='fastparquet'
)
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
columns_to_drop = [
    'VendorID',
    'lpep_dropoff_datetime',
    'store_and_fwd_flag',
    'RatecodeID',
    'PULocationID',
    'DOLocationID',
    'extra',
    'mta_tax',
    'ehail_fee',
    'improvement_surcharge',
    'total_amount',
    'payment_type',
    'trip_type',
    'congestion_surcharge',
    'tolls_amount'
]
data = data.drop(columns=columns_to_drop)
```

```python
data.info()
```

```python
# drop trips where passengers count is 0
data = data[data['passenger_count'] > 0]
```

```python
#drop trips where trip distance is 0
data = data[data['trip_distance'] > 0]
```

```python
#drop trips with not-positive fare amount
data = data[data['fare_amount'] > 0]
```

```python
# drop trips with not-positive tip amount
data = data[data['tip_amount'] >= 0]
```

```python
data.info()
```

```python
# Build new features from lpep_pickup_datetime and tip_amount
data['pickup_hour'] = data['lpep_pickup_datetime'].dt.hour
data['pickup_day_of_week'] = data['lpep_pickup_datetime'].dt.dayofweek  # 0=Monday, 6=Sunday
data['tip_applied'] = (data['tip_amount'] > 0).astype(int)  # 1 if tip given, 0 otherwise
```
